```python
import pandas as pd
import kagglehub
import os
import matplotlib.pyplot as plt
```

Dataset : download  

```python
def get_dataset():
#Download Dataset from kaggle
  path = kagglehub.dataset_download("arashnic/book-recommendation-dataset")
  #import dataset and assign to repective variable
  books = pd.read_csv(os.path.join(path,'Books.csv'))
  users = pd.read_csv(os.path.join(path,'Users.csv'))
  ratings = pd.read_csv(os.path.join(path,'Ratings.csv'))
  return (books, users, ratings)
```

Observation


```python
#Observing Datasets
def show_obs(books, users, ratings):
  #Creating a dictionary object for Datasets
  Datasets = {'books':books, 'Users':users, 'Ratings':ratings}

  for name,dataset in Datasets.items():
    print(f"{name}:\n\n", dataset.head(2), "\n")
    #Observing missing values and statistics
    print("Statistics:\n")
    dataset.info()
    print("\nNull values\n\n", dataset.isnull().astype(int).sum(), "\n\n")
    #Checking for duplicate values in datasets
    print(f"Duplicate entries in {name}: {dataset.duplicated().sum()}", '\n\n')
```

Dataset Visualisation

```python
#Plotting histogram
def plot_hist(ratings):
  plt.hist(ratings['Book-Rating'], bins=10)
  plt.title("Ratings distribution")
  plt.xlabel("Rating")
  plt.ylabel("Count")
  plt.show()

"""
Observation:
Mosty of the ratings are 0

"""
```

```python
#Box plot
def plot_box(ratings):
  plt.boxplot(ratings["Book-Rating"])
  plt.title("Box Plot of Book Ratings")
  plt.ylabel("Rating")
  plt.show()
```

Run observation and visualisation

```python
if(__name__ == "__main__"):
  (books, users, ratings) = get_dataset()
  show_obs(books, users, ratings)
  plot_hist(ratings)
  print("\n\n")
  plot_box(ratings)
```
